# Outlier Handling

```python
numeric_cols = df_clean.select_dtypes(
    include=np.number
).columns.tolist()

print(numeric_cols)
```

```python
from scipy.stats import zscore
outlier_columns = ["annual_income","loan_amount","credit_score"]
for column in outlier_columns:
    z_scores = np.abs(zscore(df_clean[column].dropna()))
    print(column,"Outliers:",(z_scores > 3).sum())
```

```python
# IQR Method

for column in outlier_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR
    outliers = df[
        (df_clean[column] < lower_limit) |
        (df_clean[column] > upper_limit)]
    print(column,"Outliers:",len(outliers))
```

```python
# Percentile Method

for column in outlier_columns:
    lower_limit = df_clean[column].quantile(0.01)
    upper_limit = df_clean[column].quantile(0.99)
    outliers = df[
        (df_clean[column] < lower_limit) |
        (df_clean[column] > upper_limit)]
    print(column,"Outliers:",len(outliers))
```

```python
df_winsorized = df_clean.copy()
for column in outlier_columns:
    lower_limit = df_winsorized[column].quantile(0.01)
    upper_limit = df_winsorized[column].quantile(0.99)
    df_winsorized[column] = np.clip(
        df_winsorized[column],lower_limit, upper_limit)

df_winsorized.head()
```

# Feature Engineering

```python
numerical_features = df_clean.select_dtypes(include=np.number).columns.tolist()
categorical_features = df_clean.select_dtypes(include="object").columns.tolist()

print("Numerical Features:")
print(numerical_features)

print("\nCategorical Features:")
print(categorical_features)
```

```python
df_clean["join_date"] = pd.to_datetime(df["join_date"],errors="coerce")

df_clean["join_date"].head()

df_clean["join_year"] = df_clean["join_date"].dt.year
df_clean["join_month"] = df_clean["join_date"].dt.month
df_clean["join_day"] = df_clean["join_date"].dt.day
df_clean["join_weekday"] = df_clean["join_date"].dt.day_name()
df_clean[[
        "join_date",
        "join_year",
        "join_month",
        "join_day",
        "join_weekday"]].head()
```

```python
# Ordinal Encoding

from sklearn.preprocessing import OrdinalEncoder
education_order = [["Secondary","Graduate","Post-Graduate"]]

ordinal_encoder = OrdinalEncoder(categories=education_order)

df_clean["education_level_encoded"] = ordinal_encoder.fit_transform(df_clean[["education_level"]])
df_clean[["education_level","education_level_encoded"]].head()
```

```python
# Label Encoding

from sklearn.preprocessing import LabelEncoder
label_encoder = LabelEncoder()

df_clean["gender_encoded"] = label_encoder.fit_transform(df_clean["gender"].astype(str))

df_clean[["gender","gender_encoded"]].head()
```

```python
# One-Hot Encoding

df_encoded = pd.get_dummies(df_clean,columns=["region","loan_purpose"],drop_first=True,dtype=int)
df_encoded.head()
```

```python
# Binning Annual Income

df_clean["income_category"] = pd.qcut(df["annual_income"],q=4,
    labels=["Low","Medium","High","Very High"],duplicates="drop")

df_clean[["annual_income","income_category"]].head()
```

```python
#Binarization

df_clean["good_credit"] = np.where(df_clean["credit_score"] > 700,1,0)
df_clean[["credit_score","good_credit"]].head()
```

```python
# Quantile Binning

from sklearn.preprocessing import KBinsDiscretizer
quantile_binner = KBinsDiscretizer(n_bins=4,encode="ordinal",strategy="quantile")

df_clean["transaction_bin"] = quantile_binner.fit_transform(df_clean[["transaction_count"]])

df_clean[["transaction_count","transaction_bin"]].head()
```

```python
# K-Means Binning

from sklearn.cluster import KMeans
kmeans = KMeans(n_clusters=3,random_state=42,n_init=10)

df["transaction_cluster"] = kmeans.fit_transform(df[["transaction_count"]])

df[["transaction_count","transaction_cluster"]].head()
```

# Feature Scalling

```python
# MaxAbsScaler

from sklearn.preprocessing import MaxAbsScaler
maxabs_scaler = MaxAbsScaler()
df_clean["loan_maxabs"] = maxabs_scaler.fit_transform(df_clean[["loan_amount"]])
df_clean[["loan_amount","loan_maxabs"]].head()
```

# Feature Construction & Transformation

```python
df_clean["spending_ratio"] = (df_clean["loan_amount"] /df_clean["annual_income"]) * 100
df_clean[["loan_amount","annual_income","spending_ratio"]].head()
```

```python
# FunctionTransformer: Log Transformation
from sklearn.preprocessing import FunctionTransformer
log_transformer = FunctionTransformer(np.log1p)
df_clean["spending_log"] = log_transformer.fit_transform(df_clean[["spending_ratio"]])
df_clean[["spending_ratio","spending_log"]].head()
```

```python
# Reciprocal Transformation
df_clean["spending_reciprocal"] = 1 / (df_clean["spending_ratio"] + 1)
df_clean[["spending_ratio","spending_reciprocal"]].head()
```

```python
# Square Root Transformation
df_clean["spending_sqrt"] = np.sqrt(df_clean["spending_ratio"].clip(lower=0))
df_clean[["spending_ratio","spending_sqrt"]].head()
```

```python
# PowerTransformer: Box-Cox

from sklearn.preprocessing import PowerTransformer
boxcox_transformer = PowerTransformer(method="box-cox")
positive_income = df_clean[["annual_income"]].dropna()
boxcox_values = boxcox_transformer.fit_transform(positive_income)
boxcox_values[:5]
```

```python
# Yeo-Johnson Transformation
yeojohnson_transformer = PowerTransformer(method="yeo-johnson")
df_clean["loan_yeojohnson"] = yeojohnson_transformer.fit_transform(df_clean[["loan_amount"]])
df_clean[["loan_amount","loan_yeojohnson"]].head()
```

```python
#ColumnTransformer

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline

numerical_columns = [
    "age",
    "annual_income",
    "loan_amount",
    "credit_score",
    "transaction_count"]

categorical_columns = [
    "gender",
    "employment_type",
    "region"]

numeric_pipeline = Pipeline(
    steps=[("imputer",SimpleImputer(strategy="median")),("scaler",StandardScaler())])

categorical_pipeline = Pipeline(steps=[("imputer",SimpleImputer(strategy="most_frequent")),
        ("encoder",OneHotEncoder(handle_unknown="ignore"))])

preprocessor = ColumnTransformer(transformers=[("num",numeric_pipeline,numerical_columns),
        ("cat",categorical_pipeline,categorical_columns)])

processed_data = preprocessor.fit_transform(df_clean)

print(processed_data.shape)
```

```python
# Debt-to-Income Ratio
df_clean["debt_to_income_ratio"] = (df_clean["loan_amount"] /df_clean["annual_income"])

df_clean[["loan_amount","annual_income","debt_to_income_ratio"]].head()
```

```python
# Average Monthly Transactions
df_clean["months_since_join"] = ((pd.Timestamp.today() -df_clean["join_date"]).dt.days / 30).clip(lower=1)

df_clean["average_monthly_transactions"] = (df_clean["transaction_count"] /df_clean["months_since_join"])

df_clean[["transaction_count","months_since_join","average_monthly_transactions"]].head()
```

```python
# Spending-to-Income Ratio

df_clean["spending_to_income_ratio"] = (df_clean["loan_amount"] /df_clean["annual_income"])
df_clean[["loan_amount","annual_income","spending_to_income_ratio"]].head()
```
